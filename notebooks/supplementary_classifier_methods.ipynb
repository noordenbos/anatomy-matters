# Supplementary table — Molecular classifier implementations

Inputs, preprocessing, assignment rules and unavailable-feature handling for four genomic and five transcriptomic classifiers. Each row links to the original method and the exact analysis-code revision.

The two tables use Great Tables and the typography and horizontal-rule conventions of `table1.ipynb`. The source is `docs/supplementary/classifier_methods.json`. No patient data or classifier execution is required. PDF, PNG and portable HTML exports are written to `figures/supplementary_classifier_methods/`.

```python
from pathlib import Path
import sys, json, importlib

REPO_ROOT = Path.cwd().resolve()
if REPO_ROOT.name == "notebooks":
    REPO_ROOT = REPO_ROOT.parent
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

import dlbcl.classifier_methods_table as methods_table
importlib.reload(methods_table)

SOURCE = REPO_ROOT / "docs/supplementary/classifier_methods.json"
OUTDIR = REPO_ROOT / "figures/supplementary_classifier_methods"
methods = json.loads(SOURCE.read_text())
assert [len(group["rows"]) for group in methods["sections"]] == [4, 5]
print("Method code revision:", methods["code_revision"])
```

## A Genomic classifiers

```python
genomic_table = methods_table.render_classifier_methods_table(methods, 0)
genomic_table
```

## B Transcriptomic classifiers

```python
transcriptomic_table = methods_table.render_classifier_methods_table(methods, 1)
transcriptomic_table
```

The combined PDF has one landscape page per modality. Individual PDF and PNG exports are suitable for figure assembly; HTML preserves the clickable code references and embeds the manuscript font.
